```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import pickle
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

```python
#Importando o dataframe que será utilizado

df = pd.read_csv('card_transdata.csv')
```

```python
#Exibindo o dataframe

df.head()
```

```python
#Verificando se existem valores ausentes

df.isnull().sum()
```

```python
#Verificando se existem registros duplicados

df.duplicated().sum()
```

```python
#Verificando a quantidade de linhas e colunas

df.shape
```

```python
#Verificando os tipos das colunas

df.dtypes
```

```python
#Verificando o balanceamento da variável target

df['fraud'].value_counts()
```

```python
#Aplicando StandardScaler para padronizar as variáveis numéricas

standard_scaler = StandardScaler()
colunas_standard_scaler = ['distance_from_home', 'distance_from_last_transaction', 'ratio_to_median_purchase_price']

for col in colunas_standard_scaler:
    df[col] = standard_scaler.fit_transform(df[[col]])
```

```python
#Exibindo o dataframe final

df.head()
```

```python
#Criando variável de features

x = df.iloc[:, 0:7].values
```

```python
#Criando variável de target

y = df.iloc[:, 7].values
```

```python
#Dividindo os dados em treino(80%) e teste(20%)

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
#Criando arquivo para salvar as variáveis de treino e teste

with open ('variaveis_train_test.pkl', mode='wb') as f:
    pickle.dump([x_train,y_train,x_test,y_test], f)
```

```python
#Criando modelo de Regressão Logística. Class_weight='balanced' devido a desbalanceamento de classes.

logistic_regression = LogisticRegression(class_weight='balanced', random_state=42)
```

```python
#Treinando o modelo

logistic_regression.fit(x_train, y_train)
```

```python
#Criando as previsões do modelo

previsoes = logistic_regression.predict(x_test)
```

```python
#Exibindo a acuracidade do modelo

accuracy_score(y_test, previsoes)
```

```python
#Exibindo e analisando as demais métricas do modelo

print(classification_report(y_test, previsoes))
```

```python
#Exibindo e analisando a matriz de confusão

print(confusion_matrix(y_test, previsoes))
```
