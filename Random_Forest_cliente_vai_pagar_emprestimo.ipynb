```python
import kagglehub
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import pickle
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
#Importando o dataframe que será utilizado

df = pd.read_csv('loan_data.csv')
```

```python
#Exibindo o dataframe

df.head()
```

```python
#Verificando se existem valores ausentes

df.isnull().sum()
```

```python
#Verificando se existem registros duplicados

df.duplicated().sum()
```

```python
#Verificando a quantidade de linhas e colunas

df.shape
```

```python
#Verificando os tipos das colunas

df.dtypes
```

```python
#Verificando o balanceamento da variável target

df['loan_status'].value_counts()
```

```python
#Selecionando somente as colunas que serão utilizadas

df = df[['person_age', 'person_education', 'person_income', 'person_emp_exp', 'person_home_ownership', 'loan_amnt', 'loan_int_rate', 'loan_percent_income', 'credit_score', 'previous_loan_defaults_on_file', 'loan_status']]
```

```python
#Aplicando LabelEncoder para transformar variáveis categóricas(strings) em valores numéricos

label_encoder = LabelEncoder()
colunas_label_encoder = ['person_education', 'person_home_ownership', 'previous_loan_defaults_on_file']

for col in colunas_label_encoder:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
#Exibindo o dataframe final

df.head()
```

```python
#Criando variável de features

x = df.iloc[:, 0:10].values
```

```python
#Criando variável de target

y = df.iloc[:, 10].values
```

```python
#Dividindo os dados em treino(80%) e teste(20%)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#Criando arquivo para salvar as variáveis de treino e teste

with open('variaveis_train_test.pkl', mode='wb') as f:
    pickle.dump([x_train, y_train, x_test, y_test], f)
```

```python
#Criando o modelo Random Forest com 150 árvores, usando entropia como critério de divisão

random_forest = RandomForestClassifier(n_estimators=150, criterion='entropy', random_state=42)
```

```python
#Treinando o modelo

random_forest.fit(x_train, y_train)
```

```python
#Criando as previsões do modelo

previsoes = random_forest.predict(x_test)
```

```python
#Exibindo a acuracidade do modelo

accuracy_score(y_test, previsoes)
```

```python
#Exibindo e analisando as demais métricas do modelo

print(classification_report(y_test, previsoes))
```

```python
#Exibindo e analisando a matriz de confusão

print(confusion_matrix(y_test, previsoes))
```
