```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pickle
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
#Importando o dataframe que será utilizado

df = pd.read_excel('bank-full.xlsx')
```

```python
#Exibindo o dataframe

df.head()
```

```python
#Verificando se existem valores ausentes

df.isnull().sum()
```

```python
#Verificando se existem registros duplicados

df.duplicated().sum()
```

```python
#Verificando a quantidade de linhas e colunas

df.shape
```

```python
#Verificando os tipos das colunas

df.dtypes
```

```python
#Verificando o balanceamento da variável target

df['y'].value_counts()
```

```python
#Removendo colunas que não serão usadas

df.drop(df.columns[[8,9,13]], axis=1, inplace=True)
```

```python
#Aplicando LabelEncoder para transformar variáveis categóricas(strings) em valores numéricos

label_encoder = LabelEncoder()
colunas_label_encoder = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'month', 'poutcome', 'y']

for col in colunas_label_encoder:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
#Aplicando StandardScaler para padronizar as variáveis numéricas

standard_scaler = StandardScaler()
colunas_standard_scaler = ['age', 'job', 'marital', 'education', 'balance', 'month', 'duration', 'campaign','previous', 'poutcome']

for col in colunas_standard_scaler:
    df[col] = standard_scaler.fit_transform(df[[col]])
```

```python
#Exibindo o dataframe final

df.head()
```

```python
#Criando variável de features

x = df.iloc[:, 0:13].values
```

```python
#Criando variável de target

y = df.iloc[:, 13].values
```

```python
#Dividindo os dados em treino(80%) e teste(20%)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#Criando arquivo para salvar as variáveis de treino e teste

with open('variaveis_train_test.pkl', mode='wb') as f:
    pickle.dump([x_train, y_train, x_test, y_test], f)
```

```python
#Criando modelo SVM com kernel rbf. Ajuste de C=1.0 para controle de margem. Class_weight='balanced' devido a desbalanceamento de classes.

svm = SVC(kernel='rbf', random_state=42, C=1.0, class_weight='balanced')
```

```python
#Treinando o modelo

svm.fit(x_train, y_train)
```

```python
#Criando as previsões do modelo

previsoes = svm.predict(x_test)
```

```python
#Exibindo a acuracidade do modelo

print(accuracy_score(y_test, previsoes))
```

```python
#Exibindo e analisando as demais métricas do modelo

print(classification_report(y_test, previsoes))
```

```python
#Exibindo e analisando a matriz de confusão

print(confusion_matrix(y_test, previsoes))
```
